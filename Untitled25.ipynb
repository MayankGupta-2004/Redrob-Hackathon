```python
import math

# Resume dataset
resumes = {
    "01": {"name": "Arjun Sharma",    "skills": "Pyhton, MachineLearning, SQL, pandas, numpy, Deep-learning"},
    "02": {"name": "Priya Nair",      "skills": "JavaScrpit, Reacts, Node.JS, MongoDb, REST api, HTML/CSS"},
    "03": {"name": "Rahul Gupta",     "skills": "Java, Spring Boot, MySql, Microservices, Docker, kubernates"},
    "04": {"name": "Sneha Patel",     "skills": "Python, TensorFlow, Keras, NLP, BERT, data-viz, matplotlib"},
    "05": {"name": "Vikram Singh",    "skills": "C++, Algoritms, Data Structure, competitive programming, python"},
    "06": {"name": "Ananya Krishnan", "skills": "javascript, vue.js, python, flask, PostgreSQL, AWS, CI/CD"},
    "07": {"name": "Karan Mehta",     "skills": "Python, Sklearn, XGboost, feature engineering, SQL, tableau"},
    "08": {"name": "Deepika Rao",     "skills": "Java, Android, Kotlin, Firebase, REST, UI/UX, figma"},
    "09": {"name": "Aditya Kumar",    "skills": "Reactjs, TypeScrpit, GraphQL, redux, tailwind, nodejs, jest"},
    "10": {"name": "Meera Iyer",      "skills": "python, R, statistics, ML, regression, clustering, Power-BI"}
}

# Job description dataset
job_descriptions = {
    "1": {"company": "Kakao", "role": "ML Engineer",
          "requiredskills": ["Python", "Machine Learning", "Deep Learning", "TensorFlow", "PyTorch", "SQL", "Data Visualization"],
          "preferredskills": ["NLP", "BERT", "Feature Engineering", "Statistics"]},
    "2": {"company": "Naver", "role": "Backend Engineer",
          "requiredskills": ["Java", "Spring Boot", "MySQL", "PostgreSQL", "Microservices", "Docker", "Kubernetes"],
          "preferredskills": ["REST API", "CI/CD", "Redis"]},
    "3": {"company": "Line", "role": "Frontend Engineer",
          "requiredskills": ["JavaScript", "React", "Vue", "TypeScript", "REST API", "HTML/CSS"],
          "preferredskills": ["Node.js", "GraphQL", "Redux", "Jest", "AWS"]}
}

# SKILL_ALIASES
SKILL_ALIASES = {
    "python": "python", "pyhton": "python",
    "java": "java",
    "javascript": "javascript", "javascrpit": "javascript", "js": "javascript",
    "typescript": "typescript", "typescrpit": "typescript",
    "c++": "cpp", "cpp": "cpp",
    "r": "r", "kotlin": "kotlin",
    "machinelearning": "machine_learning", "machine learning": "machine_learning",
    "ml": "machine_learning", "sklearn": "machine_learning",
    "deeplearning": "deep_learning", "deep learning": "deep_learning", "deep-learning": "deep_learning",
    "tensorflow": "tensorflow", "pytorch": "pytorch", "keras": "keras",
    "nlp": "nlp", "bert": "bert", "xgboost": "xgboost",
    "feature engineering": "feature_engineering",
    "statistics": "statistics", "stats": "statistics",
    "regression": "regression", "clustering": "clustering",
    "data-viz": "data_visualization", "data visualization": "data_visualization",
    "data viz": "data_visualization", "matplotlib": "data_visualization",
    "tableau": "data_visualization", "power-bi": "data_visualization",
    "power bi": "data_visualization", "powerbi": "data_visualization",
    "pandas": "pandas", "numpy": "numpy",
    "react": "react", "reacts": "react", "reactjs": "react",
    "vue": "vue", "vue.js": "vue", "vuejs": "vue",
    "redux": "redux", "tailwind": "tailwind",
    "html/css": "html_css", "html css": "html_css", "html": "html_css", "css": "html_css",
    "jest": "jest", "graphql": "graphql",
    "node.js": "nodejs", "nodejs": "nodejs", "node js": "nodejs",
    "flask": "flask",
    "spring boot": "spring_boot", "springboot": "spring_boot",
    "rest api": "rest_api", "rest": "rest_api", "restapi": "rest_api",
    "microservices": "microservices",
    "sql": "sql", "mysql": "mysql", "mysq": "mysql",
    "postgresql": "postgresql", "postgres": "postgresql",
    "mongodb": "mongodb", "redis": "redis",
    "docker": "docker",
    "kubernetes": "kubernetes", "kubernates": "kubernetes", "k8s": "kubernetes",
    "ci/cd": "ci_cd", "cicd": "ci_cd", "ci cd": "ci_cd",
    "aws": "aws", "android": "android", "firebase": "firebase",
    "algorithms": "algorithms", "algoritms": "algorithms",
    "data structure": "data_structures", "data structures": "data_structures",
    "competitive programming": "competitive_programming",
    "ui/ux": "ui_ux", "ui ux": "ui_ux", "figma": "figma"
}

# Multi-word keys sorted longest first
MULTI_WORD_KEYS = sorted([k for k in SKILL_ALIASES if ' ' in k], key=lambda x: -len(x))

# Clean and deduplicate resume skills
cleaned_resumes = {}
for rid, resume in resumes.items():
    tokens = [t.strip() for t in resume["skills"].lower().split(",")]
    cleaned_skills = []
    for token in tokens:
        matched = False
        for phrase in MULTI_WORD_KEYS:
            if phrase in token:
                cleaned_skills.append(SKILL_ALIASES[phrase])
                matched = True
                break
        if not matched:
            if token in SKILL_ALIASES:
                cleaned_skills.append(SKILL_ALIASES[token])
    # Deduplicate preserving order
    seen = set()
    deduped = []
    for s in cleaned_skills:
        if s not in seen:
            seen.add(s)
            deduped.append(s)
    cleaned_resumes[rid] = {"name": resume["name"], "skills": deduped}

# Build shared skill vocabulary
vocabulary = set()
for resume in cleaned_resumes.values():
    vocabulary.update(resume["skills"])
vocabulary = sorted(list(vocabulary))

# Compute TF-IDF for each resume
tfidf_resumes = {}
for rid, resume in cleaned_resumes.items():
    tf_idf = [0] * len(vocabulary)
    for i, skill in enumerate(vocabulary):
        if skill in resume["skills"]:
            tf = 1 / len(resume["skills"])
            df = sum(1 for r in cleaned_resumes.values() if skill in r["skills"])
            idf = math.log(10 / df)
            tf_idf[i] = tf * idf
    tfidf_resumes[rid] = tf_idf

# Build binary vectors for job descriptions
binary_jd = {}
for jid, jd in job_descriptions.items():
    binary = [0] * len(vocabulary)
    all_jd_skills = [s.lower() for s in jd["requiredskills"] + jd["preferredskills"]]
    # Normalize JD skills through SKILL_ALIASES
    normalized_jd_skills = set()
    for s in all_jd_skills:
        matched = False
        for phrase in MULTI_WORD_KEYS:
            if phrase in s:
                normalized_jd_skills.add(SKILL_ALIASES[phrase])
                matched = True
                break
        if not matched and s in SKILL_ALIASES:
            normalized_jd_skills.add(SKILL_ALIASES[s])
    for i, skill in enumerate(vocabulary):
        if skill in normalized_jd_skills:
            binary[i] = 1
    binary_jd[jid] = binary

# Compute cosine similarity and print top 3 candidates per JD
for jid, jd in job_descriptions.items():
    similarities = {}
    for resume_id, tfidf in tfidf_resumes.items():
        dot_product = sum(a * b for a, b in zip(tfidf, binary_jd[jid]))
        magnitude_tfidf = math.sqrt(sum(a ** 2 for a in tfidf))
        magnitude_binary = math.sqrt(sum(a ** 2 for a in binary_jd[jid]))
        if magnitude_tfidf * magnitude_binary == 0:
            similarity = 0
        else:
            similarity = dot_product / (magnitude_tfidf * magnitude_binary)
        similarities[resume_id] = similarity
    top_3 = sorted(similarities.items(), key=lambda x: (-x[1], cleaned_resumes[x[0]]["name"]))
    print(f"JD-{jid} - {jd['company']} ({jd['role']})")
    for i, (resume_id, similarity) in enumerate(top_3[:3]):
        print(f"{i+1}. {cleaned_resumes[resume_id]['name']} ({similarity:.2f})")
    print()
```
